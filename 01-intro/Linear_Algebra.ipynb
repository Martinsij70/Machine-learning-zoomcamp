```python
import numpy as np
```

```python
def vector_vector_mul(u, v):
    assert u.shape[0] == v.shape[0]

    n = u.shape[0]
    result = 0.0

    for i in range(n):
        result += u[i] * v[i]

    return result
```

```python
u = np.array([12, 8, 7])
```

```python
v = np.array([10, 4, 2])
```

```python
vector_vector_mul(u, v)
```

```python
u.dot(v)
```

```python
u.shape[0]
```

```python
def matrix_vector_multiplication(U, v):
    assert U.shape[1] == v.shape[0]

    num_rows = U.shape[0]
    result = np.zeros(num_rows)

    for i in range(num_rows):
        result[i] = vector_vector_mul(U[i], v)

    return result
```

```python
U = np.arange(12).reshape(3, 4)
U
```

```python
v = np.arange(4)
v
```

```python
matrix_vector_multiplication(U, v)
```

```python
U.dot(v)
```

```python
V = np.arange(12).reshape(4, 3)
V = V+1
V
```

```python
U.dot(V)
```

```python
def matrix_matrix_multiplication(U, V):
    assert U.shape[1] == V.shape[0]

    num_rows = U.shape[0]
    num_cols = V.shape[1]

    result = np.zeros((num_rows, num_cols))

    for i in range(num_cols):
        vi = V[:, i]
        uvi = matrix_vector_multiplication(U, vi)
        result[:, i] = uvi

    return result
```

```python
matrix_matrix_multiplication(U, V)
```
