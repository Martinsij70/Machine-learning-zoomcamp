**2.6 Linear Regression : Vector Form**

```python
def dot(w, xi):
    n = len(xi)

    res = 0.0
    for j in range(n):
        res += w[j] * xi[j]
        
    return res
```

```python
# putting it all in a function,

def train_linear_regression(X, y):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)

    w0 = w_full[0] 
    w = w_full[1:]

    return w0, w  
```

**2.13 Regularization**

```python
# Check note for explanation on why we had that error
```

```python
X = [
    [4, 4, 4],
    [3, 5, 5],
    [5, 1, 1],
    [5, 4, 4],
    [7, 5, 5],
    [4, 5, 5],
]
```

```python
X = np.array(X)
```

```python
XTX = X.T.dot(X)
XTX
```

```python
# The second and third columns are identical and hence, there can't be an inverse
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
# As we can see, error.
# However we don't get this error from our prepare_X function because of noisy values, i.e, the identical columns have some noisy values like .0000005, 
# etc which won't raise an error but would give large results. 
# See below
```

```python
X = [
    [4, 4, 4],
    [3, 5, 5],
    [5, 1, 1],
    [5, 4, 4],
    [7, 5, 5.0001],
    [4, 5, 5.00000001],
]
X = np.array(X)
```

```python
XTX = X.T.dot(X)
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
XTX_inv
```

```python
y = [1, 2, 3, 1, 2, 3]
```

```python
XTX_inv.dot(X.T).dot(y)
```

```python
# We can see that the values of w are large, ie, the ones where there was duplicate which is 2nd and 3rd column.
# To solve this, we add a small number to the diagonal of the matrix XTX.
# We implement this by adding an identity matrix of same shape to the matrix XTX.
```

```python
# E.g
XTX = [
    [1, 2, 2],
    [2, 1, 1.0000001],
    [2, 1.0000001, 1]
]
XTX = np.array(XTX)
```

```python
np.linalg.inv(XTX)
```

```python
# The above still has large values
# Adding .1 to the diaginal will make the result of the inv closer to normal.
XTX = [
    [1.1, 2, 2],
    [2, 1.1, 1.0000001],
    [2, 1.0000001, 1.1]
]
np.linalg.inv(np.array(XTX))
```

```python
# implement using identity matrix to add a value to the diagonal
XTX = [
    [1, 2, 2],
    [2, 1, 1.0000001],
    [2, 1.0000001, 1]
]
XTX = np.array(XTX)
```

```python
XTX = XTX + 0.01 * np.eye(3)    #the larger the value we add to the diagonal, the smaller the values on the inverse. Try with 0.1
XTX
```

```python
np.linalg.inv(XTX)
```

```python
# issue solved. No awkwardly large values
# Now applying it to our model
```

```python
def train_linear_regression_reg(X, y, r=0.001):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX = XTX + r * np.eye(XTX.shape[0])
    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)

    w0 = w_full[0] 
    w = w_full[1:]

    return w0, w  
```

```python
X_train = prepare_X(df_train)
w0, w = train_linear_regression_reg(X_train, y_train, r=0.01)

X_val = prepare_X(df_val)
y_pred = w0 + X_val.dot(w)

rmse(y_val, y_pred)
```

**2.14 Tuning the Model**

```python
# Here we discuss how to get the best value for r (parameter)
```

```python
for r in [0.00001, 0.0001, 0.001, 0.01, 0.1, 1, 10]:
    X_train = prepare_X(df_train)
    w0, w = train_linear_regression_reg(X_train, y_train, r=r)
    
    X_val = prepare_X(df_val)
    y_pred = w0 + X_val.dot(w)
    
    score =  rmse(y_val, y_pred)
    print(r, w0, score)
```
